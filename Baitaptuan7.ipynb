W7A1 : Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target)

```python
def bs(arr,left,right,target) : 
    m= (left + right) //2 
    if arr[m] == target : 
        return m 
    elif arr[m] < target : 
        return bs(arr,m+1,right,target)
    elif arr[m] > target : 
        return bs(arr,left,m -1 ,target)
    return -1 
arr = list(map(int,input().split()))
target = int(input())
print(bs(arr,0,len(arr)-1,target))
```

W7A2 : Cho danh sách arr và một
số nguyên x, đếm xem x
xuất hiện bao nhiêu lần
trong danh sách. def
count_occurrences(arr, x) ->int:



```python
def co(arr,x) : 
    for i in arr : 
        dem = 0 
        for j in arr:
            if i == j : 
                dem += 1
    return dem 
arr = list(map(int,input().split()))
x = int(input())
print(co(arr,x))
```

W7A3 : Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection
Sor

```python
def bs(arr) : 
    a = len(arr)
    for i in range(0,a) : 
        for j in range(0,a-i-1) : 
            if arr[j] > arr[j+1] :
                arr[j],arr[j+1] = arr[j+1],arr[j]
    return arr 
arr = list(map(int,input().split()))
print(bs(arr))
```

```python
def selection_sort(arr) : 
    a = len(arr)
    for i in range(0,a) :
        min = i 
        for j in range(i+1,a) : 
            if arr[min] > arr[j]:
                min = j 
        arr[i],arr[min] = arr[min],arr[i]
    return arr
arr = list(map(int,input().split()))
print(selection_sort(arr))
```

W7A4 : Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.


```python
def count(n) : 
    nhieu_nhat = 0 
    so_gan_nhat = 0
    for i in n : 
        cnt = 0 
        for j in n : 
            if i == j : 
                cnt = cnt + 1 
        if cnt > nhieu_nhat : 
            nhieu_nhat = cnt 
            so_gan_nhat = i 
    return f"{so_gan_nhat} xuat hien nhieu nhat, som nhat, {nhieu_nhat} lan" 
n = list(map(int,input().split()))
print(count(n))
```

W7A5 : Cho danh sách các số
nguyên và một số X, đếm
tất cả các cặp (a, b) sao
cho a + b = X. c


```python
def ds(n,c):
    dem = 0 
    for a in range(0,len(n)) : 
        for b in range(a+1,len(n)):
            i = (n[a]+n[b])
            if i == c : 
                dem +=1
    return dem 
n = list(map(int,input().split()))
c = int(input())
print(ds(n,c))

    
```

W7A6: Bạn được cung cấp một
danh sách các danh sách
lồng nhau, vd [[1, 2, 3], [4,
5], [6, 7, 8, 9]], hãy làm
phẳng danh sách

```python
def lp(n) : 
    kq = []
    for i in range(len(n)) : 
        for j in n[i] : 
            kq.append(j)
    return kq 
n = eval(input())
print(lp(n))
```

W7A7 : Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền kề)

```python
def nums(n) : 
    dem = 1 
    for i in range(len(n)-1): 
        if n[i] < n[i+1] : 
            dem +=1
        elif n[i] == n[i+1] : 
            dem = 1 
    return dem 
n = list(map(int,input().split()))
print(nums(n))
```

W7A9 : Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số
đếm.
Efficient solution: using
heap or binary search tree

```python
def nums(n) : 
    rs = []
    for i in range(len(n)-1):
        dem = 0 
        for j in range(i+1,len(n)) : 
            if n[i] > n[j] :
                dem +=1
        rs.append(dem)
    rs.append(0)
    return rs
n = list(map(int,input().split()))
print(nums(n))
        
```
