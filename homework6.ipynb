W6A16 Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n, m = map(int, input().split())

a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)
for i in range(n):
    for j in range(m):
        print(f'{a[i][j]:>4}', end='')
    print()
```

W6A17 Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input())
a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)
for i in range(n):
    print(a[i][i], end=' ')
print()

for i in range(n):
    print(a[i][n - 1 - i], end=" ")
```

W6A18 Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
n, m, k = map(int, input().split())

a = [list(map(int, input().split())) for _ in range(n)]
k -= 1
print(sum(a[k]))
print(sum(a[i][k] for i in range(n)))
```

W6A19 Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
n = list(map(int, input().split()))
max_value = max(n)
for i in range(len(n)):
    if n[i] == max_value:
        print(i, end=' ')
```

W6A20 Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1


```python
n = list(map(int, input().split()))
k = int(input())
if k in n:
    print(n.index(k))
else:
    print(-1)
```

W6A21 Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện


```python
n = list(input().split())
d = {}
for i, v in enumerate(n):
    if v not in d:
        d[v] = []
    d[v].append(i)
for k in d:
    d[k] = tuple(d[k])
print(d)
```

W6A24 Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 0 2 3 2 4 9


```python
n = list(map(int, input().split()))
d = []
d1 = []
for i in range(len(n)):
    if n[i] == 0:
        d.append(n[i])
    elif n[i] != 0:
        d1.append(n[i])
print(d + d1)
```

W6A26 Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
a = list(map(int, input().split()))

total = 0

for i in range(1, len(a) - 1):
    if a[i] > a[i - 1] and a[i] > a[i + 1]:
        total += a[i]

print(total)
```
